## EXTRACT DATA

```python
# RUN THIS CELL ONLY TO UPDATE THE ISTAT DATA
# this cell save the istat data in a raw data folder in order to avoid running the api call every time
url = "https://esploradati.istat.it/SDMXWS/rest/data/41_983"
header = {"Accept":"application/vnd.sdmx.data+csv;version=1.0.0"}
response = requests.get(url, headers=header)
print(response.status_code)

with open("data/raw/istat_incidenti.csv", "w", encoding="utf-8") as f:
    f.write(response.text)
```

```python
# RUN THIS CELL ONCE TO DOWNLOAD SITUAS DATA
# this cell save the situas data from 31-12-2001 to 31-12-2024 in 24 different csv files
date_ranges = [f"{year}-12-31" for year in range(2001, 2025)]

# to download files in csv format is necessary to set json and header parameters.
# You can acknowledge this by using dev tools on the web page at the below specified url.
json_situas = {"orderFields":[],"orderDirects":[],"pFilterFields":[],"pFilterValues":[]}
header_situas = {"Content-Type": "application/json-patch+json"}

for i in date_ranges:
    url_situas = f"https://situas.istat.it/ShibO2Module/api/Report/Spool/{i}/74?&pdoctype=CSV"
    time.sleep(1)
    response_situas = requests.post(url_situas, json = json_situas, headers = header_situas)
    print(response_situas.status_code)
    if response_situas.status_code == 200:
        with open(f"data/raw/situas_{i[0:4]}.csv", "w", encoding="utf-8") as f:
            f.write(response_situas.text)
    else:
        # if an http status code different from 200 is encounterd, this make sure that cell execution is stopped
        raise SystemExit(f"Abort due to http error {response_situas.status_code}")
print(f"All done - {len(date_ranges)} files downloaded")
```

## CLEAN DATA

We begin cleaning data with `SITUAS data` csv file

```python
situas_df = pd.read_csv("data/raw/situas_2001.csv", sep = ";", decimal = ",")
situas_df["Anno"] = 2001
print(f"Year 2001:{situas_df.shape}")
situas_df = [situas_df]

for i in range(2002,2025):
    tmp_df = pd.read_csv(f"data/raw/situas_{i}.csv", sep = ";", decimal = ",")
    tmp_df["Anno"] = i # add a column with the year of the data

# Analyzing the csv situas data, it can be noticed that from 2015 on there is another column 
# not present in the previous years "Codice Provincia (Storico)"". This is simply the code that 
# in the previous years is called "Codice Provincia/Uts" which from 2015 on contains different values. 
# So, we can drop "Codice Provincia/Uts" and change "Codice Provincia (Storico)"" name in "Codice Provincia/Uts"
# in order to obtain a uniform dataframe.
    if i in range(2015,2025):
        tmp_df.drop(columns=["Codice Provincia/Uts"], inplace = True)
        tmp_df.rename(columns={"Codice Provincia (Storico)": "Codice Provincia/Uts"}, inplace = True)

    # check if columns name are the same among csv files in both directions
    check_column_names_sx = set(situas_df[0].columns) - set(tmp_df.columns)
    check_column_names_dx = set(tmp_df.columns) - set(situas_df[0].columns)

    # check if columns order are the same among csv files
    check_column_order = list(situas_df[0].columns) == list(tmp_df.columns)

    if len(check_column_names_sx) == 0 and len(check_column_names_dx) == 0 and check_column_order:
        print(f"Year {i}: {tmp_df.shape}")
        situas_df.append(tmp_df)
    else:
        print(situas_df[0].columns)
        print(tmp_df.columns)
        raise SystemExit(f"Abort due to column name mismatch in situas_{i}.csv")
situas_df = pd.concat(situas_df, ignore_index = True)

```

From the shapes of the previous cell, we can notice that the number of rows change for every situas csv file. Since that every record in these csv files is a different town, this means that the number of town changes every year. This happens because some towns splits or merge toghether with others. This kind of town usually are very small so it's difficult that they could be interesting for investment from business. So we can simply keep just the towns that are present in every csv file. In order to confirm this strategy, we can calculate how many towns are in common between all csv files.

```python
# Determine the total number of distinct years covered by the dataset.
tot_years = situas_df["Anno"].nunique()

# Count the number of years in which each municipality appears.
count_years_per_town = situas_df.groupby("Codice Comune (alfanumerico)")["Anno"].nunique()

# Select towns that are present in all years of the dataset.
towns_common_code = count_years_per_town[count_years_per_town == tot_years].index
print(len(towns_common_code))
```

Since that the majority of towns don't change through years, we can assess that the bigger towns are not excluded from the analysis and we can confirm the strategy. We drop changing towns because to do a correct analysis, we have to work on towns that have the same amount of data 

```python
# Keep only towns that are present in every year of the dataset,
# then reset the DataFrame index after filtering.
situas_df = situas_df[
    situas_df['Codice Comune (alfanumerico)'].isin(towns_common_code)
]
situas_df.reset_index(inplace=True, drop=True)
print(situas_df.info())
situas_df.head(10)
```

```python
# drop the column "Comune (dizione straniera) since that have a lot nan values and is not useful for our analysis"
situas_df.drop(columns=['Comune (dizione straniera)'],inplace=True)
```

```python
# Loading the situas data, it can be noticed that the first column name is not correct. 
# It is called "ï»¿Codice Ripartizione geografica" instead of "Codice Ripartizione geografica". 
# This is due to a problem with the encoding of the csv file. We can fix this by renaming the column.
situas_df.rename(columns={'ï»¿Codice Ripartizione geografica': 'Codice Ripartizione geografica'}, inplace=True)
print(situas_df.info())
print(situas_df.sample(10))
```

From the previous cell we can see there are some nan values (about 2000) in the column "Sigla automobilistica".

```python
print(situas_df[pd.isna(situas_df['Sigla automobilistica'])])
print(situas_df['Codice Regione'][pd.isna(situas_df['Sigla automobilistica'])].unique())
```

Investigating on this columns, we can see that all nan values have "Codice Regione" $= 15$. This code stands for "Campania". We can fill these nan with the correct code using the column "Codice Provincia"

```python
# we can discover the province code for each of the 5 province of Campania.
# Then we can fill nan values in column "Sigla automobilistica" with the correct string code.
province_code = []
province_name = []
for i in range(0,len(situas_df)):
    if situas_df['Comune'].iloc[i] in ['Napoli','Caserta','Salerno','Avellino','Benevento']:
        province_code.append(situas_df['Codice Provincia/Uts'].iloc[i])
        province_name.append(situas_df['Comune'].iloc[i])
province_campania_df = pd.DataFrame({'Codice Provincia/Uts': province_code, 'Nome Provincia/Uts': province_name})
print(province_campania_df)

for i in range(0,len(situas_df)):
    if pd.isna(situas_df['Sigla automobilistica'].iloc[i]):
        if situas_df['Codice Provincia/Uts'].iloc[i] == 61:
            situas_df.loc[i,'Sigla automobilistica'] = 'CE'
        elif situas_df['Codice Provincia/Uts'].iloc[i] == 62:
            situas_df.loc[i,'Sigla automobilistica'] = 'BN'
        elif situas_df['Codice Provincia/Uts'].iloc[i] == 63:
            situas_df.loc[i,'Sigla automobilistica']  = 'NA'
        elif situas_df['Codice Provincia/Uts'].iloc[i] == 64:
            situas_df.loc[i,'Sigla automobilistica'] = 'AV'
        elif situas_df['Codice Provincia/Uts'].iloc[i] == 65:
            situas_df.loc[i,'Sigla automobilistica'] = 'SA'

print(situas_df.info())
print(situas_df.sample(10))
```

From the output we can notice that it's necessary to investigate further on the meaning of each variable

After doing some online research at the following links:

https://esploradati.istat.it/SDMXWS/rest/datastructure/IT1/DCIS_INCIDMORFER_COM/
https://esploradati.istat.it/SDMXWS/rest/codelist/IT1/CL_TIPO_DATO22

you can understand the meaning of each variable. First, it is clear that each row represents a single observation, whose value is identified by the “OBS_VALUE” column, which is labeled by various categorical dimensions identified by the columns: “DATA_TYPE” and “RESULT,” which have the labels KILLINJ/ROADACC and F/M/9, respectively (F stands for injured, M for fatality, and 9 is the label for “total,” i.e., the total number of accidents that occurred in a given municipality over the course of a year).
So, for example, if you have a record with a structure like this:\

REF_AREA = 001001\
DATA_TYPE = ROADACC\
RESULT = 9\
TIME_PERIOD = 2001\
OBS_VALUE = 10\

this means that in the town with code 001001, a total of 10 traffic accidents occurred in 2001

Given the structure of the data frame, it is advisable to apply a pivot transformation to create columns showing the actual number of injuries, deaths, and accidents for each town and for each year. This will facilitate the linking of the ISTAT data with the SITUAS data.

```python
# pivot trasformation
istat_tmp_df = istat_df.pivot_table(index = ['REF_AREA','TIME_PERIOD'], columns = ['DATA_TYPE','RESULT'], values = 'OBS_VALUE').reset_index()

# Rename columns to more meaningful names
istat_tmp_df.columns = ['Ref_Area','Time_Period','N_Injured','N_Dead','N_Accidents']

# Insert "Dataflow" and "Frequency" columns lost in the pivoting operation
istat_tmp_df.insert(0, 'Dataflow', istat_df['DATAFLOW'])
istat_tmp_df.insert(1, 'Frequency', istat_df['FREQ'])
istat_df = istat_tmp_df

# Convert numeric columns to int
istat_df['N_Injured'] = istat_df['N_Injured'].astype(int)
istat_df['N_Dead'] = istat_df['N_Dead'].astype(int)
istat_df['N_Accidents'] = istat_df['N_Accidents'].astype(int)
istat_df.head(10)
```

```python
# after the pivoting of the table, we can check if there are any missing values
print(istat_df.info())
```

Now, we can join the dataframes (situas and istat) into a bigger and enriched dataframe containing information about car accidents that happened in each town in Italy each year from 2001 to 2024 (istat data). Moreover, this dataframe will have information about towns, such as their population and their surface area (situas data)

```python
merged_df = pd.merge(istat_df, situas_df, left_on = ['Ref_Area','Time_Period'], right_on = ['Codice Comune (alfanumerico)','Anno'], how = 'inner')
print(merged_df.info())
print(merged_df.sample(10))
```

From merged_df.info() we notice some errors that has to be handled:
1. "Comune" column has some null values. We can check why there are these null and decide to keep them as is, drop or fill them.
2. "Superficie (Kmq)" datatype is string and not int or float. We have to convert it in order to create other columns useful for the analysis.
3. There are redondant columns such as "Ref_Area" - "Codice Comune (alfanumerico)" - "Codice Comune (numerico)" and "Time Period" - "Anno". We can keep just one for year and one for town code.

```python
# check towns code of towns that have null values in the "Comune" column after the merge operation.
merged_df[merged_df['Comune'].isnull()]
```

From the previous cell output it can be noticed that just one town has null values in the column "Comune" in the dataset. Since that we know only province and state of this town, it's not possibile to reveal its name. Moreover, null values are very few (24 vs ~170.000) so we can drop these records.

```python
# drop rows with null values in the "Comune" column and reset the index of the dataframe
merged_df.drop(merged_df[merged_df['Comune'].isnull()].index, inplace=True)
merged_df.reset_index(inplace=True, drop=True)
print(merged_df.info())
print(merged_df.sample(10))
```

```python
# Replace the dot in the "Superficie (Kmq)" column with an empty string to remove thousands separators
merged_df['Superficie (Kmq)'] = merged_df['Superficie (Kmq)'].str.replace('.','')

# Replace the comma in the "Superificie (Kmq)" column with a dot to convert it to a float
merged_df['Superficie (Kmq)'] = merged_df['Superficie (Kmq)'].str.replace(',','.').astype('float64')

merged_df.info()
```

```python
# Drop redondant columns for town code
merged_df.drop(['Ref_Area', 'Codice Comune (alfanumerico)'], inplace=True, axis=1)

# Drop redondant columns for year
merged_df.drop(['Anno'], inplace=True,axis=1)

# Rename "Codice Comune (numerico)" column to "Codice Comune"
merged_df.rename(columns={'Codice Comune (numerico)' : 'Codice Comune'}, inplace=True)

merged_df.info()
```

Now, we can enrich the dataframe with other columns where we compute some useful statistics such as popultaion density and rates about accidents, deaths, injuries and fatalities. These columns will provide useful data to make business decision about investments.

Before adding columns, it's good practice to check if there are any zeros in the columns used as denominators in rate calculations.

```python
# check if there are any rows with 0 values in the "Popolazione residente" and "Superficie (Kmq)" columns
print((merged_df['Popolazione residente'] == 0).sum())
print((merged_df['Superficie (Kmq)'] == 0).sum())
print((merged_df['N_Accidents'] == 0).sum())
```

Since that some little towns in specific years have 0 registered accidents, we have to deal with division by 0 problem in these row. We will insert nan values in the column "Fatality_Rate (%)" for these rows to indicate a lack of data (not registered accidents).

```python
# round the values in the "Superficie (Kmq)" column to 2 decimal places
merged_df['Superficie (Kmq)'] = merged_df['Superficie (Kmq)'].round(2)

# add "Population_Density (Pop/Kmq)" column to the merged dataframe
merged_df['Population_Density (Pop/Kmq)'] = (merged_df['Popolazione residente'] / merged_df['Superficie (Kmq)']).round(2)

# add "Accident_Rate_1000_Pop" column to the merged dataframe
merged_df['Accident_Rate_1000_Pop'] = ((merged_df['N_Accidents'] / merged_df['Popolazione residente']) * 1000).round(2)

# add "Death_Rate_1000_Pop" column to the merged dataframe
merged_df['Death_Rate_1000_Pop'] = ((merged_df['N_Dead'] / merged_df['Popolazione residente'])* 1000).round(2)

# add "Injury_Rate_1000_Pop" column to the merged dataframe
merged_df['Injury_Rate_1000_Pop'] = ((merged_df['N_Injured'] / merged_df['Popolazione residente']) * 1000).round(2)

# add "Fatality_Rate (%)" column to the merged dataframe. If the number of accidents is 0, 
# the fatality rate will be set to NaN to avoid division by zero and to comunicate clearly lack of data.
merged_df['Fatality_Rate (%)'] = (np.where(merged_df['N_Accidents'] == 0, np.nan, (merged_df['N_Dead'] / merged_df['N_Accidents']) * 100)).round(2)

# add "Accident_Density (Acc/Kmq)" column to the merged dataframe
merged_df['Accident_Density (Acc/Kmq)'] = (merged_df['N_Accidents'] / merged_df['Superficie (Kmq)']).round(2)

# convert "Popolazione legale" column to integer
merged_df['Popolazione legale'] = merged_df['Popolazione legale'].astype('int64')

print(merged_df.info())
print(merged_df.sample(10))
```

## EDA (Exploratory Data Analysis) 

### UNIVARIATE ANALYSIS

Now that we have cleaned istat and situas data, we can move to exploratory data analysis where we can study the correlation between variables and the distribution of data.

```python
# create a table with descriptive statistics for the "N_Accidents", "N_Dead", and "N_Injured" columns
# and add to it the interquartile range, skewness, and kurtosis
desc_counts = merged_df[['N_Accidents', 'N_Dead', 'N_Injured']].describe()

iqr_counts = desc_counts.loc['75%'] - desc_counts.loc['25%']
skew_counts = merged_df[['N_Accidents', 'N_Dead', 'N_Injured']].skew()
kurt_counts = merged_df[['N_Accidents', 'N_Dead', 'N_Injured']].kurt()
iqr_counts.name = 'iqr'
skew_counts.name = 'skewness'
kurt_counts.name = 'kurtosis'

stats_table_counts = pd.concat([desc_counts, iqr_counts.to_frame().T, skew_counts.to_frame().T, kurt_counts.to_frame().T])

stats_table_counts
```

```python
# set the figure layout to 2 rows and 3 columns
fig, axes = plt.subplots(2, 3, figsize=(20, 12))

# specify the variables to plot
variables = ['N_Accidents', 'N_Dead', 'N_Injured']

# loop over the variables and plot the boxplots and histograms
for i, var in enumerate(variables):
    sns.boxplot(data=merged_df, x=var, ax=axes[0, i], color = "#0077FF")
    axes[0, i].set_xscale('log')
    axes[0, i].set_title(f'Boxplot of {var} (log scale)')

# exclude zeros from the histogram plots because kde parameter do not apply automatically log scale parameter to values to handle the log(0) issue
    sns.histplot(data=merged_df[merged_df[var] > 0], x=var, kde=True, ax=axes[1, i], log_scale=True, color = "#0077FF")
    axes[1, i].set_title(f'Distribution of {var} (log scale)')

plt.show()
```

From the graphs and statistics shown above, it can be seen that `"N_Accidents"`, `"N_Dead"`, and `"N_Injured"` contain a large number of extreme values (which are not true outliers, as they are all plausible values). This is due to the fact that Italy is a country characterized by a very high number of small towns and relatively few large population centers. This asymmetry necessitated the use of a logarithmic scale rather than a linear one, since for all three variables the majority of the values (fortunately, I might add) are clustered near zero, with very high extreme values associated with very large towns that “stretch” the data distribution, creating a very long right tail. This can be seen both in the very low IQR and in the extremely high skewness and kurtosis values, indicating an extremely asymmetric, right-skewed data distribution. It is also interesting to note the complete absence of the "box" in the boxplot for `"N_Dead"`, a phenomenon caused by the fact that both Q1 and Q3 are zero, indicating that at least 75% of the values for the number of deaths are 0.

```python
# create a table with descriptive statistics for the "Population_Density (Pop/Kmq)", "Accident_Density (Acc/Kmq)" columns
# and add to it the interquartile range, skewness, and kurtosis
desc_densities = merged_df[['Population_Density (Pop/Kmq)', 'Accident_Density (Acc/Kmq)']].describe()

iqr_densities = desc_densities.loc['75%'] - desc_densities.loc['25%']
skew_densities = merged_df[['Population_Density (Pop/Kmq)', 'Accident_Density (Acc/Kmq)']].skew()
kurt_densities = merged_df[['Population_Density (Pop/Kmq)', 'Accident_Density (Acc/Kmq)']].kurt()
iqr_densities.name = 'iqr'
skew_densities.name = 'skewness'
kurt_densities.name = 'kurtosis'

stats_table_densities = pd.concat([desc_densities, iqr_densities.to_frame().T, skew_densities.to_frame().T, kurt_densities.to_frame().T])

stats_table_densities
```

```python
# set the figure layout to 2 rows and 2 columns
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# specify the variables to plot
variables = ['Population_Density (Pop/Kmq)', 'Accident_Density (Acc/Kmq)']

# loop over the variables and plot the boxplots and histograms
for i, var in enumerate(variables):
    sns.boxplot(data=merged_df, x=var, ax=axes[0, i], color = "#0077FF")
    axes[0, i].set_xscale('log')
    axes[0, i].set_title(f'Boxplot of {var} (log scale)')

# exclude zeros from the histogram plots because kde parameter do not apply automatically log scale parameter to values to handle the log(0) issue
    sns.histplot(data=merged_df[merged_df[var] > 0], x=var, kde=True, ax=axes[1, i], log_scale=True, color = "#0077FF")
    axes[1, i].set_title(f'Distribution of {var} (log scale)')

plt.show()
```

Once again, the contrast between small towns with few accidents and large towns with many accidents results in a highly positively-skewed data distribution with a long right tail, although the distribution here is less skewed than in the previous case. This again led to the use of a logarithmic scale, to avoid crowding all the values on the left and severely limiting the readability of the graphs. It is interesting to note that the distribution of the variable `"Population_Density (Pop/Kmq)"`, when using a logarithmic scale, closely approximates a normal (unstandardized) distribution, confirming that demographic variables often follow a log-normal distribution. The `"Accident_Density (Acc/Kmq)"` variable exhibits completely different behavior: it appears to be multimodal, suggesting that accident density is influenced by factors other than population density alone, such as tourism or the presence of major, heavily trafficked roads.

```python
# create a table with descriptive statistics for the "Accident_Rate_1000_Pop", "Death_Rate_1000_Pop", "Injury_Rate_1000_Pop" columns
# and add to it the interquartile range, skewness, and kurtosis
desc_rates = merged_df[['Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop']].describe()

iqr_rates = desc_rates.loc['75%'] - desc_rates.loc['25%']
skew_rates = merged_df[['Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop']].skew()
kurt_rates = merged_df[['Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop']].kurt()
iqr_rates.name = 'iqr'
skew_rates.name = 'skewness'
kurt_rates.name = 'kurtosis'

stats_table_rates = pd.concat([desc_rates, iqr_rates.to_frame().T, skew_rates.to_frame().T, kurt_rates.to_frame().T])

stats_table_rates
```

```python
# set the figure layout to 2 rows and 3 columns
fig, axes = plt.subplots(2, 3, figsize=(20, 12))

# specify the variables to plot
variables = ['Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop']

# loop over the variables and plot the boxplots and histograms
for i, var in enumerate(variables):
    sns.boxplot(data=merged_df, x=var, ax=axes[0, i], color = "#0077FF")
    axes[0, i].set_xscale('log')
    axes[0, i].set_title(f'Boxplot of {var} (log scale)')

# exclude zeros from the histogram plots because kde parameter do not apply automatically log scale parameter to values to handle the log(0) issue
    sns.histplot(data=merged_df[merged_df[var] > 0], x=var, kde=True, ax=axes[1, i], log_scale=True, color = "#0077FF")
    axes[1, i].set_title(f'Distribution of {var} (log scale)')

plt.tight_layout()
plt.show()
```

Once again, the descriptive statistics show a skewed distribution across all three variables (positively skewed with a long right tail), with the `"Death_Rate_1000_Pop"` variable being much more skewed than the other two (here too, the box is empty because at least 75% of the values in the dataset are 0). This is likely because normalizing by population does not eliminate the "rarity" of the "death" event; on the contrary, small towns with even just one death have a very high per-capita rate, contributing to the large number of extreme values. The variables `"Injury_Rate_1000_Pop"` and `"Accident_Rate_1000_Pop"`, unlike the previously examined variables "N_Accidents" and "N_Injured" (and unlike "Accident_Density" as well), are much more regular (unimodal) and symmetrical, closely resembling a non-standardized normal distribution. This is likely because normalizing by population removes some of the geographic variability, isolating a phenomenon that depends more on a municipality's population than on its geography.

### BIVARIATE ANALYSIS

Moving on to the bivariate analysis, given that—as we saw in the univariate analysis section—the distributions of all numerical variables are highly skewed, it is appropriate to compare `Pearson` and `Spearman` correlation matrices, since the former is much more sensitive to the presence of extreme values and outliers because it assumes linear relationships between the variables

```python
# calculate the Pearson and Spearman correlation matrices
corr_pearson = merged_df[['N_Accidents','N_Dead', 'N_Injured', 'Population_Density (Pop/Kmq)', 'Accident_Density (Acc/Kmq)', 'Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop', 'Fatality_Rate (%)']].corr(method = 'pearson')
corr_spearman = merged_df[['N_Accidents','N_Dead', 'N_Injured', 'Population_Density (Pop/Kmq)', 'Accident_Density (Acc/Kmq)', 'Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop', 'Fatality_Rate (%)']].corr(method = 'spearman')

# display the correlation matrices on heatmaps
sns.heatmap(corr_pearson.round(2), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Pearson Correlation Matrix')
plt.show()

sns.heatmap(corr_spearman.round(2), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Spearman Correlation Matrix')
plt.show()
```

As one might expect, the two correlation matrices are quite different from one another. The most notable case is the combination of `"Fatality_Rate (%)"` and `"Death_Rate_1000_Pop"`, which shows a very high correlation according to Spearman but a much lower one according to Pearson. This is likely because the relationship between the two variables is monotonic but not linear, and Pearson's correlation is not well suited to capturing this kind of relationship. It is also worth noting that, in raw counts, `"N_Accidents"`, `"N_Dead"`, and `"N_Injured"` are strongly correlated with one another, likely because, as the number of accidents increases, so does the number of potential injuries and deaths. Another interesting relationship is that between `"Accident_Rate_1000_Pop"` and `"N_Accidents"`, which is well captured by Spearman's correlation but not by Pearson's, likely because a nonlinear relationship exists between the two (for example, in large towns—that is, at extreme values—the behavior may differ from that in smaller ones, as large town may have a very high number of accidents but a comparatively low per-capita rate). This difference between the two matrices quantitatively confirms the strong asymmetry already observed in the univariate analysis.

## CLUSTERING

Since the purpose of the analysis is to help a company operating in the field of risk prevention identify the best cities to invest in, it is useful to perform a clustering analysis to identify high-risk, medium-risk, and low-risk towns. 

In addition to accident rate, death rate, and injury rate, we will also use the fatality rate as a variable, since a company that deals with risk prevention and road traffic management is interested not only in the frequency of events but also in their severity.

Unlike in exploratory analysis, where we retained NaN to avoid confusing “no data” with “zero risk,” here for clustering we choose to impute 0, since for the purposes of risk classification, a town with no recorded incidents still represents a very low-risk case, consistent with the segmentation objective

```python
# create a copy of the original dataframe to not alter the original
clustering_features = merged_df[['Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop', 'Fatality_Rate (%)']].copy()

# fill missing values with 0
clustering_features['Fatality_Rate (%)'] = clustering_features['Fatality_Rate (%)'].fillna(0)
```

Given what was observed in the EDA—namely, that all the variables to be used in the clustering are highly asymmetric—a logarithmic transformation is applied to make the data distribution more symmetric. This prevents extreme values from interfering with the proper functioning of the K-means algorithm, which minimizes the distance of each data point from its cluster centroid.\
If this transformation were not applied, it would result in a very small cluster containing all the extreme values (for example, those of very large towns) and a huge cluster containing all the other towns.

```python
# create a copy of the extracted columns to not alter the original extracted ones
clustering_features_log = clustering_features.copy()

# apply log1p transformation to deal with the log(0) issue for little towns that have zero accidents.
# If not done, the log(0) will be -inf and the clustering will fail.
for col in clustering_features_log.columns:
    clustering_features_log[col] = np.log1p(clustering_features_log[col])
```

```python
# calculate the skewness of each variable and put them in a dataframe
pd.DataFrame({'variables' : ['Accident_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Injury_Rate_1000_Pop', 'Fatality_Rate (%)'],
              'Skewness' : clustering_features_log.skew(),}).reset_index(drop=True)
```

The logarithmic transformation has significantly reduced the skewness of all variables; Death_Rate_1000_Pop remains the most skewed of the four, consistent with the inherent rarity of the “death” event (which therefore contains numerous values equal to 0) even after the transformation—a known limitation that will, however, be mitigated by the subsequent standardization.

The four variables chosen for clustering, after the logarithmic transformation, still have different scales. If used as is, the algorithm would assign greater weight to the variable with the highest range—due more to a numerical artifact than to conceptual importance. `Standardization`—which sets the mean of all distributions to 0 and the standard deviation to 1—places all variables on an equal numerical footing, so that each contributes equally to the calculation of distances.

```python
# scale the data with StandardScaler
scaler = StandardScaler()
clustering_features_scaled = scaler.fit_transform(clustering_features_log)

# create a dataframe with the scaled data
clustering_features_scaled_df = pd.DataFrame(clustering_features_scaled, columns = clustering_features_log.columns, index = clustering_features_log.index)
```

```python
clustering_features_scaled_df.describe()
```

The description of the variables confirms that the standardization was successful (mean ~0 and standard deviation ~1). However, even after standardization, most values of the `"Death_Rate_1000_Pop"` variable are identical or very close to each other, while the maximum reaches approximately 24 standard deviations above the mean. The "rare event" nature of this variable is therefore not eliminated by the logarithmic transformation and standardization; it is merely made more manageable for the K-means algorithm.

To determine the number of clusters for risk class classification, we can use the `elbow method` rather than choosing an arbitrary number of clusters

```python
# set WCSS to an empty list
WCSS = []
# create a range of values for the number of clusters
k_range = range(1, 11)

# calculate the WCSS for each value of k
for k in k_range:
    kmeans_elbow = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_elbow.fit(clustering_features_scaled_df)
    WCSS.append(kmeans_elbow.inertia_)

# plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_range, WCSS, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.xticks(k_range)
plt.show()
```

The `“elbow method”` graph clearly shows that the WCSS decreases rapidly between $k=1$ and $k=3$. From $k=3$ onward, its rate of decrease slows significantly. This behavior clearly highlights the “elbow” at $k=3$, suggesting the use of three risk classes: `Low Risk`, `Medium Risk`, and `High Risk`. 

```python
# create a KMeans model
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# combine the training and label assignment 
cluster_labels = kmeans.fit_predict(clustering_features_scaled_df)

# add the cluster labels to the original dataframe so risk lables will be linked to towns' information
merged_df['Town_Risk_Cluster (numeric)'] = cluster_labels
```

Since the k-means algorithm itself does not provide “high,” “medium,” and “low” risk labels (as one might expect), to determine which cluster represents a given risk class, you need to visualize the mean values of the variables used by algorithm

```python
merged_df.groupby('Town_Risk_Cluster (numeric)')[['Injury_Rate_1000_Pop', 'Death_Rate_1000_Pop', 'Accident_Rate_1000_Pop', 'Fatality_Rate (%)']].mean()
```

* `Cluster 0` has low values across the board, thus identifying the `“low risk”` category.
* `Cluster 1` has intermediate values for all variables, thus identifying the `“medium risk”` category.
* `Cluster 2` has values similar to cluster 1 in terms of accidents/injuries but significantly higher for deaths/fatality rates. This cluster therefore represents the `“high-risk”` category.\
 It is interesting to note that this last cluster is not distinguished by a higher number of accidents but by significantly greater severity. This confirms that including the variables “Fatality_Rate (%)” and “Death_Rate_1000_Pop” was crucial for clearly distinguishing this cluster from the other two.



```python
# create a list to store the risk levels
Town_risk_levels = []

# assign risk levels to each cluster based on the cluster number
for cluster in merged_df['Town_Risk_Cluster (numeric)']:
    if cluster == 0:
        Town_risk_levels.append('Low Risk')
    elif cluster == 1:
        Town_risk_levels.append('Medium Risk')
    elif cluster == 2:
        Town_risk_levels.append('High Risk')

merged_df['Town_Risk_Cluster (label)'] = Town_risk_levels
```
